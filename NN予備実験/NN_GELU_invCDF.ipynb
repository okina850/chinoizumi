```python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
from utils.target_functions import *
from scipy.special import ndtri

f = fast_inv_F_Y_given_X

# ==========================================
# 1. カスタム有理関数活性化関数の定義 (Rational Activation)
# ==========================================
class LearnableRationalActivation(nn.Module):
    def __init__(self):
        super().__init__()
        # 初期値はGELUやReLUの形に無理に似せず、まずは安定して恒等写像（y=x）に近くなるように設定
        self.a = nn.Parameter(torch.tensor([0.0, 1.0, 0.0, 0.0], dtype=torch.float32))
        self.b = nn.Parameter(torch.tensor([0.0, 0.0], dtype=torch.float32))

    def forward(self, x):
        # 分子 P(x)
        P = self.a[0] + self.a[1] * x + self.a[2] * (x ** 2) + self.a[3] * (x ** 3)
        
        # 【修正】分母 Q(x) が絶対にゼロ以下にならないよう、xの項に絶対値を噛ませる
        # これにより、bがどんな値になろうとも Q(x) >= 1.0 が保証され、爆発（極の発生）を防ぎます
        Q = 1.0 + torch.abs(self.b[0]) * torch.abs(x) + torch.abs(self.b[1]) * (x ** 2)
        
        return P / Q

# 2. データ生成
grid_size = 50
x_coord = np.linspace(1e-6, 1 - 1e-6, grid_size)
y_coord = np.linspace(-4.0, 4.0, grid_size)
X_mesh, Y_mesh = np.meshgrid(x_coord, y_coord, indexing='ij')

x_flat = X_mesh.ravel()
y_flat = Y_mesh.ravel()

# x を probit 空間に変換
x_probit = ndtri(x_flat)

z_true = f(X_mesh, Y_mesh)
z_flat = z_true.ravel()

features = torch.tensor(np.stack([x_probit, y_flat], axis=1), dtype=torch.float32)
labels = torch.tensor(z_flat, dtype=torch.float32).view(-1, 1)

# ==========================================
# 3. モデル定義 (Rational 4層モデル)
# ==========================================
class Rational_MLP_4Layer(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(2, 64),
            LearnableRationalActivation(),
            nn.Linear(64, 64),
            LearnableRationalActivation(),
            nn.Linear(64, 64),
            LearnableRationalActivation(),
            nn.Linear(64, 64),
            LearnableRationalActivation(),
            nn.Linear(64, 1)
        )
    def forward(self, x):
        return self.net(x)

model = Rational_MLP_4Layer()

# 有理関数のパラメータ（a, b）も同時に最適化するため、通常通りAdamに渡す
optimizer = optim.Adam(model.parameters(), lr=0.002)
# scheduler = optim.lr_scheduler.MultiStepLR(
#     optimizer, milestones=[2000, 4000], gamma=0.1
# )

criterion = nn.MSELoss()

# 4. 訓練ループ
print("Rational Activationを用いた学習を開始します...")

for epoch in range(5000):
    optimizer.zero_grad()
    pred = model(features)
    loss = criterion(pred, labels)
    
    loss.backward()
    optimizer.step()
    scheduler.step()
    
    if (epoch + 1) % 500 == 0:
        print(f"Epoch {epoch+1}: Loss {loss.item():.6e} | LR {optimizer.param_groups[0]['lr']:.6f}")

print("訓練が完了しました。")

# 5. 近似用ラッパー関数
def f_approx(x, y):
    x_probit_eval = ndtri(np.array(x).ravel())
    coords = torch.tensor(np.stack([x_probit_eval, np.array(y).ravel()], axis=1), dtype=torch.float32)
    with torch.no_grad():
        res = model(coords).numpy().reshape(np.shape(x))
    return res

# 6. 誤差解析
n_eval = 51
x_eval = np.linspace(1e-6, 1 - 1e-6, n_eval)
y_eval = np.linspace(-4.0, 4.0, n_eval)
X_eval, Y_eval = np.meshgrid(x_eval, y_eval, indexing='ij')

f_eval = f(X_eval, Y_eval)
f_eval_approx = f_approx(X_eval, Y_eval)

abs_err = np.abs(f_eval - f_eval_approx)
rel_err = abs_err / np.max(np.abs(f_eval))
print(f"Max Relative Error: {np.max(rel_err):.6f}")

# 7. 可視化
fig = plt.figure(figsize=(14, 6))
ax1 = fig.add_subplot(1, 2, 1, projection='3d')
ax1.plot_surface(X_eval, Y_eval, f_eval, cmap='viridis', alpha=0.8)
ax1.set_title("True Function")

ax2 = fig.add_subplot(1, 2, 2, projection='3d')
ax2.plot_surface(X_eval, Y_eval, f_eval_approx, cmap='plasma', alpha=0.8)
ax2.set_title("NN with Rational Activation")

fig_err = plt.figure(figsize=(7, 6))
ax3 = fig_err.add_subplot(1, 1, 1, projection='3d')
ax3.plot_surface(X_eval, Y_eval, rel_err, cmap='inferno', alpha=0.8)
ax3.set_title("Relative Error (Rational)")

plt.show()
```

```python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
from utils.target_functions import *
from scipy.special import ndtri

f = fast_inv_F_Y_given_X

# ==========================================
# 1. Floater-Hormann活性化関数の定義
# ==========================================
class FloaterHormannActivation(nn.Module):
    def __init__(self, num_nodes=15, d=3):
        super().__init__()
        self.num_nodes = num_nodes
        self.d = d
        
        # 補間を行う基準となるノード（チェビシェフノードの配置を初期値にする）
        # 近似領域（probit空間の主要な範囲として一旦 -4.0 から 4.0 を想定）
        nodes_init = 4.0 * np.cos((2 * np.arange(num_nodes) + 1) * np.pi / (2 * num_nodes))
        self.nodes = nn.Parameter(torch.tensor(nodes_init, dtype=torch.float32))
        
        # 各ノードにおける関数値（これがNNのウェイトと同様に学習によって最適化される）
        values_init = nodes_init.copy()  # 初期値は恒等写像 y = x に近くする
        self.values = nn.Parameter(torch.tensor(values_init, dtype=torch.float32))
        
        # Floater-Hormannの重み w_i の事前計算（符号が交互に反転することで極を排除する）
        w = np.zeros(num_nodes)
        for i in range(num_nodes):
            sum_sign = 0
            for k in range(max(0, i - d), min(i, num_nodes - 1 - d) + 1):
                # 組み合わせ数の計算
                from scipy.special import comb
                prod = 1.0
                for j in range(k, k + d + 1):
                    if j != i:
                        prod /= abs(nodes_init[i] - nodes_init[j])
                sum_sign += comb(d, i - k) * prod
            w[i] = ((-1) ** (i - d)) * sum_sign
            
        # 重みの絶対値（符号は一意に決まるため、大きさだけをパラメータ化、または固定）
        self.w = torch.tensor(w, dtype=torch.float32)

    def forward(self, x):
        # x: (batch_size, num_features)
        # 重心形式有理内挿の計算: 
        # R(x) = sum( (w_i * f_i) / (x - x_i) ) / sum( w_i / (x - x_i) )
        
        # 計算の安定化のために次元を拡張してブロードキャスト
        # x_diff: (batch_size, num_features, num_nodes)
        x_expanded = x.unsqueeze(-1)
        nodes_expanded = self.nodes.view(1, 1, -1)
        
        # 分母のゼロ割（x がちょうどノード上の点になったとき）を避けるためのeps
        diff = x_expanded - nodes_expanded
        eps = 1e-6
        inv_diff = 1.0 / (diff + torch.sign(diff) * eps)
        
        # 重みと逆差分の積
        w_expanded = self.w.view(1, 1, -1).to(x.device)
        v_expanded = self.values.view(1, 1, -1)
        
        # 重心形式の分子と分母の計算
        num = torch.sum(w_expanded * v_expanded * inv_diff, dim=-1)
        den = torch.sum(w_expanded * inv_diff, dim=-1)
        
        return num / den

# 2. データ生成
grid_size = 50
x_coord = np.linspace(1e-6, 1 - 1e-6, grid_size)
y_coord = np.linspace(-4.0, 4.0, grid_size)
X_mesh, Y_mesh = np.meshgrid(x_coord, y_coord, indexing='ij')

x_flat = X_mesh.ravel()
y_flat = Y_mesh.ravel()
x_probit = ndtri(x_flat)

z_true = f(X_mesh, Y_mesh)
z_flat = z_true.ravel()

features = torch.tensor(np.stack([x_probit, y_flat], axis=1), dtype=torch.float32)
labels = torch.tensor(z_flat, dtype=torch.float32).view(-1, 1)

# ==========================================
# 3. モデル定義 (Floater-Hormann 4層モデル)
# ==========================================
class FH_MLP_4Layer(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(2, 64),
            FloaterHormannActivation(num_nodes=15, d=3),
            nn.Linear(64, 64),
            FloaterHormannActivation(num_nodes=15, d=3),
            nn.Linear(64, 64),
            FloaterHormannActivation(num_nodes=15, d=3),
            nn.Linear(64, 64),
            FloaterHormannActivation(num_nodes=15, d=3),
            nn.Linear(64, 1)
        )
    def forward(self, x):
        return self.net(x)

model = FH_MLP_4Layer()

# 恒等写像に近い初期値から始めるため、最初の歩幅は少し小さめ (0.001) に設定
optimizer = optim.Adam(model.parameters(), lr=0.002)
# scheduler = optim.lr_scheduler.MultiStepLR(
#     optimizer, milestones=[2000, 4000], gamma=0.1
# )

criterion = nn.MSELoss()

# 4. 訓練ループ
print("Floater-Hormann Activationを用いた学習を開始します...")

for epoch in range(5000):
    optimizer.zero_grad()
    pred = model(features)
    loss = criterion(pred, labels)
    
    loss.backward()
    optimizer.step()
    scheduler.step()
    
    if (epoch + 1) % 500 == 0:
        print(f"Epoch {epoch+1}: Loss {loss.item():.6e} | LR {optimizer.param_groups[0]['lr']:.6f}")

print("訓練が完了しました。")

# 5. 近似用ラッパー関数
def f_approx(x, y):
    x_probit_eval = ndtri(np.array(x).ravel())
    coords = torch.tensor(np.stack([x_probit_eval, np.array(y).ravel()], axis=1), dtype=torch.float32)
    with torch.no_grad():
        res = model(coords).numpy().reshape(np.shape(x))
    return res

# 6. 誤差解析
n_eval = 51
x_eval = np.linspace(1e-6, 1 - 1e-6, n_eval)
y_eval = np.linspace(-4.0, 4.0, n_eval)
X_eval, Y_eval = np.meshgrid(x_eval, y_eval, indexing='ij')

f_eval = f(X_eval, Y_eval)
f_eval_approx = f_approx(X_eval, Y_eval)

abs_err = np.abs(f_eval - f_eval_approx)
rel_err = abs_err / np.max(np.abs(f_eval))
print(f"Max Relative Error: {np.max(rel_err):.6f}")

# 7. 可視化
fig = plt.figure(figsize=(14, 6))
ax1 = fig.add_subplot(1, 2, 1, projection='3d')
ax1.plot_surface(X_eval, Y_eval, f_eval, cmap='viridis', alpha=0.8)
ax1.set_title("True Function")

ax2 = fig.add_subplot(1, 2, 2, projection='3d')
ax2.plot_surface(X_eval, Y_eval, f_eval_approx, cmap='plasma', alpha=0.8)
ax2.set_title("NN with Floater-Hormann Activation")

fig_err = plt.figure(figsize=(7, 6))
ax3 = fig_err.add_subplot(1, 1, 1, projection='3d')
ax3.plot_surface(X_eval, Y_eval, rel_err, cmap='inferno', alpha=0.8)
ax3.set_title("Relative Error (Floater-Hormann)")

plt.show()
```
